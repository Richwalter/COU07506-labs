```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
n = np.arange(-5, 11)

delta = (n == 0).astype(float)
step  = (n >= 0).astype(float)
ramp  = step * n
expo  = 0.8 ** step
```

```python
fig, axes = plt.subplots(1, 4, figsize=(13, 2.6))

for ax, x, title in zip(
    axes,
    [delta, step, ramp, expo],
    ["δ(n)", "u(n)", "r(n)", "(0.8)^n u(n)"]
):
    ax.stem(n, x)
    ax.set_title(title)
    ax.set_xlabel("n")

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

n = np.arange(-5, 11)
u = (n >= 0).astype(float)

a_values = [0.5, -0.8, 1.1]

fig, axes = plt.subplots(1, 3, figsize=(12, 3))

for ax, a in zip(axes, a_values):
    x = (a ** n) * u

    ax.stem(n, x)
    ax.set_title(f"a = {a}")
    ax.set_xlabel("n")

plt.tight_layout()
plt.show()
```

```python
n = np.arange(0, 41)

x1 = np.cos(2*np.pi*n/8)
x2 = np.cos(2*np.pi*n/7.5)

fig, axes = plt.subplots(2, 1, figsize=(10, 6))

axes[0].stem(n, x1)
axes[0].set_title("cos(2πn/8)")

axes[1].stem(n, x2)
axes[1].set_title("cos(2πn/7.5)")

plt.tight_layout()
plt.show()
```

```python
from scipy.io import wavfile
import numpy as np
import matplotlib.pyplot as plt
import os
```

```python
fname = "walter.wav"

if os.path.exists(fname):
    fs, x = wavfile.read(fname)

    if x.ndim > 1:
        x = x.mean(axis=1)

    x = x / np.max(np.abs(x))

else:
    fs = 16000
    t = np.arange(0, 2, 1/fs)
    x = 0.5*np.sin(2*np.pi*440*t)

n = np.arange(len(x)) / fs

print("Sampling rate =", fs)
print("Duration =", len(x)/fs)

plt.plot(n, x)
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.title("My recording")
plt.show()
```
